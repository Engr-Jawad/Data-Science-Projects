```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# top ten high risk observation 
top_ten = df.nlargest(10,"Near_Miss_Score")
print("Top 10 highest-risk observations:")
print(top_ten)
print("__"*35)

categorical_column = ["Hour","Weather","Traffic_Level","Vehicle_Type","Road_Type"]
for col in categorical_column:
    print(top_ten[col].value_counts())
print("__"*35)



```

# Finding  the Most Dangerous Combination

```python
# city + weather
city_weather = df.groupby(["City","Weather"])["Near_Miss_Score"].mean()
print(city_weather)
```

```python
# Road type + Weather
road_type_weather = df.groupby(["Road_Type", "Weather"])["Near_Miss_Score"].mean()
road_type_weather
```

```python
# traffic level + weather 
traffic_level_weather = df.groupby(["Traffic_Level", "Weather"])["Near_Miss_Score"].mean()
traffic_level_weather
```

```python
# Road type + traffic level 
road_type_traffic_level = df.groupby(["Road_Type","Traffic_Level"])["Near_Miss_Score"].mean()
road_type_traffic_level
```
