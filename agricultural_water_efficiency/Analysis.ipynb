# Basic analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/agricultural_water_efficiency/agricultural_water_efficiency_10000.csv")
```

```python
# which region occure the most in the data ?
most_occuring_region = df["Region"].mode()
print("most occuring region in the data " ,most_occuring_region)

# highest temprature 
high_temp = df["Temperature_C"].max()
print("high temperature in celcius : ",high_temp)

# lowest temperature 
low_Temp = df["Temperature_C"].min()
print("lowest Temprature in Celcius : ", low_Temp)
print("__"*45)

# high temperature Region 
idx = df["Temperature_C"].idxmax()
print("Region with highest temperature:", df.loc[idx, "Region"])
print("Highest temperature:", df.loc[idx, "Temperature_C"])
print("__" * 45)
# Average temperature 
avg_temp = df["Temperature_C"].mean()
print("Average temperature : ", avg_temp)

# lowest temprature recorded area 
idx = df["Temperature_C"].idxmin()
print("Region with the lowest temperature : ",df.loc[idx,"Region"])
print("lowest temperature : ",df.loc[idx,"Temperature_C"])

```

```python
# average rainfall 
avg_rainfall = df["Rainfall_mm"].mean()
print("Average rainfall in milli meter : ", avg_rainfall.round(4))

# highest rainfall in milli meter 
high_rainfall_mm = df["Rainfall_mm"].max()
print("highest rainfall in milli meter : ", high_rainfall_mm)

# lowest rainfall in milli meter 
low_Rainfall_mm = df["Rainfall_mm"].min()
print("lowest rainfall in milli meter : ", low_Rainfall_mm)
print("__"* 45)

# rainfall by region 
rainfall_by_reg = df.groupby("Region")["Rainfall_mm"].mean()
print("\nRainfall by region :\n", rainfall_by_reg,4)
print("__" * 45)

# Average irrigation 
avg_irrigation = df["Irrigation_mm"].mean()
print("\nAverage Irrigation : ", avg_irrigation)

# high irrigation in milli meter
high_irrigation = df["Irrigation_mm"].max()
print("high irrigation in milli meter : ", high_irrigation)

# low irrigation in milli meter 
low_irrigation = df["Irrigation_mm"].min()
print("low irrigation in milli meter : ",low_irrigation)
print("__" * 45)
# high  irrigation recorded region 
idx = df["Irrigation_mm"].idxmax()
print("\nRegion recorded high irrigation in millimeter : ",df.loc[idx,"Region"])
print("highest irrigation in milli meter : ", df.loc[idx,"Irrigation_mm"])

# lowest irrigation recorded region 
idx = df["Irrigation_mm"].idxmin()
print("\nlowest irrigation recorded region : ", df.loc[idx,"Region"])
print("lowest irrigation in milli meter : ", df.loc[idx,"Irrigation_mm"])
```

```python
# average of the water used in liters columns 
avg_water_used = df["Water_Used_Liters"].mean()
print("Average water used in liters : ", avg_water_used)

# median of water used in liters 
median_water_used = df["Water_Used_Liters"].median()
print("Median of water consumption : ", median_water_used)

# maximum water used in liters 
max_water_used = df["Water_Used_Liters"].max()
print("Maximum water used in liters : ",max_water_used)

# minimum water used in liter
min_water_used = df["Water_Used_Liters"].min()
print("Minimum water used in liters : ",min_water_used)
print("__"* 45)

# In which region used the most water used in liters 
idx = df["Water_Used_Liters"].idxmax()
print("\nRegion in which most water used in liters : ", df.loc[idx,"Region"])
print("How many liters water used in peshawar : ", df.loc[idx,"Water_Used_Liters"])

# In which region used the minimum water 
idx = df["Water_Used_Liters"].idxmin()
print("Region in which minumum water is used in liters : ", df.loc[idx,"Region"])
print("How many liters water used in Dir : ",df.loc[idx,"Water_Used_Liters"])
```

```python
# average area hectares
avg_area_hectares = df["Area_Hectares"].mean()
print("Average of the Area Hectares : ", avg_area_hectares.round(2))

# maximum area 
max_area = df["Area_Hectares"].max()
print("Maximum Area in Hectares : ", max_area)

# minimum area 
min_area = df["Area_Hectares"].min()
print("Minimum area in Heactares : ", min_area)



```

```python
# Detecting outliers in the temprature columns 
outliers = []
def detect_outliers(df):
    threshold = 3 
    mean = np.mean(df)
    std = np.std(df)
    for i in df:
        Z_score = (i - mean)/ std
        if np.abs(Z_score > threshold):
            outliers.append(i)
    return outliers
outliers_in_Temperature_C = detect_outliers(df["Temperature_C"])
print("there are no outliers in the Temperatures Columns : ",outliers)
```

```python
# Detecting outliers in the Rainfall column
outliers = []
def detect_outliers(df):
    threshold = 3 
    mean = np.mean(df)
    std = np.std(df)
    for i in df:
        Z_score = (i - mean) / std 
        if Z_score > threshold : 
            outliers.append(i)  
    return outliers
outliers_in_Rainfall = detect_outliers(df["Rainfall_mm"])
print("Outliers in the rainfall data  :",outliers) 
print("total outliers in the rainfall data is : ",np.size(outliers_in_Rainfall))

    
```
