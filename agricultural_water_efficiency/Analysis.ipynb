# Detecting outliers in the columns 

```python
df.head(2)
```

```python
# Detecting outliers in the temprature columns 
outliers = []
def detect_outliers(df):
    threshold = 3 
    mean = np.mean(df)
    std = np.std(df)
    for i in df:
        Z_score = (i - mean)/ std
        if np.abs(Z_score > threshold):
            outliers.append(i)
    return outliers
outliers_in_Temperature_C = detect_outliers(df["Temperature_C"])
print("there are no outliers in the Temperatures Columns : ",outliers)
print("__"*50)

# Detecting outliers in the Rainfall column using Z score 
outliers_in_Rainfall = detect_outliers(df["Rainfall_mm"])
print("\nOutliers in the rainfall data  :",outliers) 
print("total outliers in the rainfall data is : ",np.size(outliers_in_Rainfall))
print("__"*50)


# detecting outliers in the  Irrigation_mm 
outlier_in_irrigation = detect_outliers(df["Irrigation_mm"])
print("\noutliers in the  irrigation data : ",outlier_in_irrigation)
print("total number of outliers in the irrigation data : ",np.size(outlier_in_irrigation))
print("__"*50)



# detecting outliers in the water used in liters  
outliers_in_water_used = detect_outliers(df["Water_Used_Liters"])
print("\noutliers in the water used in liters data : ", outliers_in_water_used)
print("Total number of outliers in water used in liters : ",np.size(outliers_in_water_used))
print("__"*50)


# detecting outliers in the Area Hectares
outlier_in_Area = detect_outliers(df["Area_Hectares"])
print("\noutliers in the  Area Hectares data : ",outlier_in_Area)
print("Total number of outliers in Area Hectares : ", np.size(outlier_in_Area))
print("__"*50)


# detecting outliers in the Crop yeild tons 
outlier_in_Crop_Yield_Tons = detect_outliers(df["Crop_Yield_Tons"])
print("\noutliers in the  Crop_Yield_Tons data : ",outlier_in_Crop_Yield_Tons)
print("Total number of outliers in Crop_Yield_Tons : ", np.size(outlier_in_Crop_Yield_Tons))
print("__"*50)


# detecting outliers in the soil moisture percent 
outlier_in_Soil_Moisture_Percent = detect_outliers(df["Soil_Moisture_Percent"])
print("\noutliers in the  Soil_Moisture_Percent : ",outlier_in_Soil_Moisture_Percent)
print("Total number of outliers in Soil_Moisture_Percent : ", np.size(outlier_in_Soil_Moisture_Percent))
print("__"*50)



```

# removing the outliers from the data set 

```python
# Removing the outliers from the Rainfall data 
Q1 = df["Rainfall_mm"].quantile(0.25)
Q3 = df["Rainfall_mm"].quantile(0.75)
IQR = Q3 - Q1
print("Inter Quartile Range : ", IQR.round(3))
lower_bound = Q1 - (1.5* IQR)
print("Lower bound : ",lower_bound)
upper_bound = Q3 + (1.5 * IQR)
print("Upper bound : ",upper_bound)

outlier_Mask = ((df["Rainfall_mm"] < lower_bound)| (df["Rainfall_mm"] > upper_bound))
print("Total Outliers in Rainfall data : ", outlier_Mask.sum())

df_cleaned = df.loc[~outlier_Mask].copy()
# Compare results
print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
print("Rows removed:", len(df) - len(df_cleaned))
```

# Removing the outliers from the whole data set 

```python

def remove_outliers(df):
    numeric_cols = df.select_dtypes(include=np.number).columns

    outlier_mask = pd.Series(False, index=df.index)

    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)

        col_outliers = (
            (df[col] < lower_bound) |
            (df[col] > upper_bound)
        )

        print(f"{col}: {col_outliers.sum()} outliers")

        outlier_mask = outlier_mask | col_outliers

    df_cleaned = df.loc[~outlier_mask].copy()

    print("\nOriginal rows:", len(df))
    print("Cleaned rows:", len(df_cleaned))
    print("Rows removed:", outlier_mask.sum())

    return df_cleaned

df_cleaned = remove_outliers(df)
```
