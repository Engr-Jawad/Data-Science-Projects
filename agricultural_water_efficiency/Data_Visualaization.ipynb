```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/agricultural_water_efficiency/cleaned_agricultural_data.csv")
df
```

# Compare average yield per hectare by crop.

```python
df["Yeild_Per_Hectares"] = (df["Crop_Yield_Tons"] / df["Area_Hectares"])
avg_yeild = df.groupby("Crop_Yield_Tons")["Area_Hectares"].mean()
plt.figure(figsize=(8,8))
plt.bar(avg_yeild.index,avg_yeild.values)
plt.title("Average Crop Yield per Hectare",fontsize = 20)
plt.xlabel("Crop Type")
plt.ylabel("Yield (Tons per Hectare)")

plt.tight_layout()
plt.savefig("crop_yield_per_Hectares.png",bbox_inches = "tight",dpi= 300)
plt.show()
```
