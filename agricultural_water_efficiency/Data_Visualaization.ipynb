# Data Visualaization  

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/agricultural_water_efficiency/cleaned_agricultural_data.csv")
df
```

# Compare average yield per hectare by crop.

```python
df["Yeild_Per_Hectares"] = (df["Crop_Yield_Tons"] / df["Area_Hectares"])
avg_yeild = df.groupby("Crop_Yield_Tons")["Area_Hectares"].mean()
plt.figure(figsize=(8,8))
plt.bar(avg_yeild.index,avg_yeild.values)
plt.title("Average Crop Yield per Hectare",fontsize = 20)
plt.xlabel("Crop Type")
plt.ylabel("Yield (Tons per Hectare)")

plt.tight_layout()
plt.savefig("crop_yield_per_Hectares.png",bbox_inches = "tight",dpi= 300)
plt.show()
```

# Create a histogram showing the distribution of water consumption.

```python
water_consumption = df["Water_Used_Liters"]
plt.Figure(figsize=(9,6))
plt.hist(water_consumption,bins = 10,color = "blue",edgecolor = "black",label="Water_Comsumption")
plt.title("Distribution of Water Consumption",fontsize = "20")
plt.legend()
plt.xlabel("Water Consumption")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("Distribution_of_Water_Consumption.png",dpi= 300 ,bbox_inches = "tight")
plt.show()
```

```python
# Investigate rainfall versus yield and calculate correlation.
plt.figure(figsize=(9,6))
plt.scatter(df["Rainfall_mm"],df["Crop_Yield_Tons"], alpha = 0.5,color = "blue",marker= '^')
plt.title("Rainfall VS Crop Yeild",fontsize = 20)
plt.xlabel("Crop yeild")
plt.ylabel("Rainfall MM")
plt.tight_layout()
plt.savefig("Rainfall_VS_Crop_yeild.png",dpi =300,bbox_inches = "tight")
plt.show()
```

```python
# correlation 
correlation = df["Rainfall_mm"].corr(df["Crop_Yield_Tons"])
print(correlation)
# very weak positive linear relationship between rainfall and crop yield per tons in your dataset.
```

```python
# Temprature distribution 
temp = df["Temperature_C"].dropna()
plt.figure(figsize=(8,7))
plt.hist(temp,bins=10,color="black",edgecolor = "blue")
plt.title("Distribution of Agricultural Temperatures")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("Agricultural_temprature_distribution.png",dpi =300,bbox_inches = "tight")
plt.show()

# the temprature data is normall distributed 
```

```python
# Compare crop yield distributions and inspect potential outliers.
plt.figure(figsize=(8,7))
plt.boxplot(df["Crop_Yield_Tons"],vert=False)
plt.title("Box Plot of the Crop Yeild Tons",fontsize = 20)
plt.show()
```

```python
# Analyze monthly average yield trends
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month

average_yeild_trends = df.groupby("Month")["Crop_Yield_Tons"].mean()
plt.figure(figsize=(8,8))
plt.plot(average_yeild_trends,color = "Red",marker = "o")
plt.title("Analyze monthly average yield trends",fontsize = 22)
plt.xlabel("Month")
plt.ylabel("Crop_Yield_Tons")
plt.grid(linestyle="-",color = "Black")
plt.tight_layout()
plt.savefig("Analyze_monthly_average_yield.png",dpi = 300,bbox_inches = "tight")
plt.show()
```
