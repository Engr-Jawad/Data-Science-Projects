# Data Visualaization  

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/agricultural_water_efficiency/cleaned_agricultural_data.csv")
df
```

# Compare average yield per hectare by crop.

```python
df["Yeild_Per_Hectares"] = (df["Crop_Yield_Tons"] / df["Area_Hectares"])
avg_yeild = df.groupby("Crop_Yield_Tons")["Area_Hectares"].mean()
plt.figure(figsize=(8,8))
plt.bar(avg_yeild.index,avg_yeild.values)
plt.title("Average Crop Yield per Hectare",fontsize = 20)
plt.xlabel("Crop Type")
plt.ylabel("Yield (Tons per Hectare)")

plt.tight_layout()
plt.savefig("crop_yield_per_Hectares.png",bbox_inches = "tight",dpi= 300)
plt.show()
```

# Create a histogram showing the distribution of water consumption.

```python
water_consumption = df["Water_Used_Liters"]
plt.Figure(figsize=(9,6))
plt.hist(water_consumption,bins = 10,color = "blue",edgecolor = "black",label="Water_Comsumption")
plt.title("Distribution of Water Consumption",fontsize = "20")
plt.legend()
plt.xlabel("Water Consumption")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("Distribution_of_Water_Consumption.png",dpi= 300 ,bbox_inches = "tight")
plt.show()
```

```python
# Investigate rainfall versus yield and calculate correlation.
plt.figure(figsize=(9,6))
plt.scatter(df["Rainfall_mm"],df["Crop_Yield_Tons"], alpha = 0.5,color = "blue",marker= '^')
plt.title("Rainfall VS Crop Yeild",fontsize = 20)
plt.xlabel("Crop yeild")
plt.ylabel("Rainfall MM")
plt.tight_layout()
plt.savefig("Rainfall_VS_Crop_yeild.png",dpi =300,bbox_inches = "tight")
plt.show()
```
