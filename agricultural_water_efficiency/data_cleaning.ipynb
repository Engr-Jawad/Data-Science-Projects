```python
# total duplicate value in the data set 
sum_of_dup = df.duplicated().sum()
print("\nsum of the duplicated rows in the data set : ",sum_of_dup)
print("__"*45)


# tota null value in the data set 
total_null_value = df.isnull().sum()
print("\ntotal null value in the data set : \n", total_null_value)
print("__"*45)

# filling the null value in the data set 
df["Region"] = df["Region"].fillna(df["Region"].mode()[0])
df["Crop_Type"] = df["Crop_Type"].fillna(df["Crop_Type"].mode()[0])
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Irrigation_mm"] = df["Irrigation_mm"].fillna(df["Irrigation_mm"].median())



# checking the data types of the data 
data_types = df.dtypes
print("\nData types of the columns : \n",data_types)
print("__"* 45)

# converting the data type of the date column into correct formate 
df["Date"] = pd.to_datetime(df["Date"])
print("Data type of the date columns : ",df["Date"].dtypes)
print("__"* 45)

# droping duplicates rows from the data set 
df = df.drop_duplicates()







```

```python
df.to_csv("clean_data.csv",index = False)
```
