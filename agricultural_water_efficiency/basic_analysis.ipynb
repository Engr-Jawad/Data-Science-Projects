```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/agricultural_water_efficiency/agricultural_water_efficiency_10000.csv")
```

```python
df.head()
```

```python
# which region occure the most in the data ?
most_occuring_region = df["Region"].mode()
print("most occuring region in the data " ,most_occuring_region)

# highest temprature 
high_temp = df["Temperature_C"].max()
print("high temperature in celcius : ",high_temp)

# lowest temperature 
low_Temp = df["Temperature_C"].min()
print("lowest Temprature in Celcius : ", low_Temp)
print("__"*45)

# high temperature Region 
high_temp_reg = df.groupby("Region")["Temperature_C"].max()
print("\nHigh tehmperature region : ",high_temp_reg )
print("__" * 45)
# Average temperature 
avg_temp = df["Temperature_C"].mean()
print("Average temperature : ", avg_temp)

```

```python
# average rainfall 
avg_rainfall = df["Rainfall_mm"].mean()
print("Average rainfall in milli meter : ", avg_rainfall.round(4))
# highest rainfall in milli meter 
high_rainfall_mm = df["Rainfall_mm"].max()
print("highest rainfall in milli meter : ", high_rainfall_mm)
# lowest rainfall in milli meter 
low_Rainfall_mm = df["Rainfall_mm"].min()
print("lowest rainfall in milli meter : ", low_Rainfall_mm)
# rainfall by region 
rainfall_by_reg = df.groupby("Region")["Rainfall_mm"].max()
rainfall_by_reg
```
