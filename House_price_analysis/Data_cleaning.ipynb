# Data Cleaning 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/home/jawad/my_jupyter_env/Data_Science_Projects/House_price_analysis/house_prices_realistic_dirty.csv")
df.head()
```

```python
# total duplicates in the data set 
total_duplicates = df.duplicated().sum()
print("Total duplicates values in the data set : ",total_duplicates)

# total null values in the data set 
total_null = df.isnull().sum()
print("\nTotal null values in the data set : \n",total_null)

# droping the duplicates rows 
df = df.drop_duplicates()
print("\nduplicates values are removed from the data set : ",df.duplicated().sum())





```

```python
# filling the null values in the data set 
df["Area_sqft"] = df["Area_sqft"].fillna(df["Area_sqft"].median())
df["Bedrooms"] =df["Bedrooms"].fillna(df["Bedrooms"].median())
df["Bathrooms"] = df["Bathrooms"].fillna(df["Bathrooms"].median())

df["Age_years"] = df["Age_years"].fillna(df["Age_years"].median())
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
df["Garage"] = df["Garage"].fillna(df["Garage"].mode()[0])
df["Crime_Rate"] = df["Crime_Rate"].fillna(df["Crime_Rate"].median())
df["Distance_Center_km"] = df["Distance_Center_km"].fillna(df["Distance_Center_km"].median())
df["Price_PKR"] = df["Price_PKR"].fillna(df["Price_PKR"].median())
print("After filling the null values are equall to : ",df.isnull().sum())
```

```python
# checking the data types
data_types = df.dtypes
print("Data types : \n",data_types)
print("--"*34)

# converting the bedrooms data type from float to int 
df["Bedrooms"] = df["Bedrooms"].astype("Int32")
df["Bathrooms"] = df["Bathrooms"].astype("Int32")
df["Age_years"] = df["Age_years"].astype("Int32")

print("\nAfter the correction data types of the data  are : \n",df.dtypes)
```

# standardize the location and Grage columns

```python
df["Location"] = df["Location"].str.strip()
df["Location"] = df["Location"].str.lower()
df["Location"] = df["Location"].str.title()
df["Garage"] = df["Garage"].str.strip()
df["Garage"] = df["Garage"].str.lower()
df["Garage"] = df["Garage"].str.title()
df

```

# correcting invalid values

```python
columns = ["Area_sqft","Bedrooms","Bathrooms","Floors","Age_years","Crime_Rate","Distance_Center_km","Price_PKR"]
for i in columns:
    df[i] = np.where(df[i] < 1 ,1,df[i])

df
```

# Detecting outliers

```python
def detect_outlier(df,columns):
    Q1 = df[columns].quantile(0.25)
    Q3 = df[columns].quantile(0.75)
    Inter_quartile_range = Q3 - Q1
    lower_bound = Q1 - 1.5 * Inter_quartile_range
    upper_bound = Q3 + 1.5 * Inter_quartile_range
    outliers = df[(df[columns] < lower_bound) | (df[columns] > upper_bound)]
    print("Q1 : ",Q1)
    print("Q3 : ",Q3)
    print("Inter quartile range : ",Inter_quartile_range)
    print("lower bound : ",lower_bound)
    print("upper bound : ",upper_bound)
    print("total outlers : ",len(outliers))
    return outliers

# outliers in the area sqft column
area_sqft_outliers = detect_outlier(df,"Area_sqft")
print("__"*34)
# outlies in the Distance_Center_km
outlies_Distance_Center_km = detect_outlier(df,"Distance_Center_km")
print("__"*34)
# outliers in Price_PKR
outliers_Price_PKR = detect_outlier(df,"Price_PKR")

```
